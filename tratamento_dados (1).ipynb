```python
#20. CRIANDO A BASE TRATADA


# Converter as colunas para o tipo data
dados["data_inicio"] = pd.to_datetime(
    dados["data_inicio"],
    errors="coerce",
    dayfirst=True
)

dados["data_fim"] = pd.to_datetime(
    dados["data_fim"],
    errors="coerce",
    dayfirst=True
)

# Verificar os tipos de dados
print(dados[["data_inicio", "data_fim"]].dtypes)

# Visualizar as datas convertidas
display(dados[["data_inicio", "data_fim"]].head())

# Converter as datas para o formato brasileiro
dados["data_inicio"] = pd.to_datetime(
    dados["data_inicio"], errors="coerce", dayfirst=True
).dt.strftime("%d/%m/%Y")

dados["data_fim"] = pd.to_datetime(
    dados["data_fim"], errors="coerce", dayfirst=True
).dt.strftime("%d/%m/%Y")

# Visualizar as datas
display(dados[["data_inicio", "data_fim"]].head())

# Converter as colunas para números inteiros
dados["carga_horaria"] = pd.to_numeric(
    dados["carga_horaria"], errors="coerce"
).astype("Int64")

dados["matriculas_aptas"] = pd.to_numeric(
    dados["matriculas_aptas"], errors="coerce"
).astype("Int64")

# Verificar os tipos de dados
print(dados[["carga_horaria", "matriculas_aptas"]].dtypes)

# Visualizar os dados
display(dados[["carga_horaria", "matriculas_aptas"]].head())


```

```python
#Criando uma cópia dos dados tratados

dados_tratados = dados.copy()

dados_tratados.to_csv( "dados_tratados.csv", index=False)
```

```python
#21. CONFERÊNCIA FINAL

print("=" * 45)
print("CONFERÊNCIA ANTES DO TRATAMENTO")
print("=" * 45)

print("Quantidade de registros:", dados.shape[0])
print("Quantidade de colunas:", dados.shape[1])
print("Total de valores ausentes:",
      dados.isnull().sum().sum())
print("Total de linhas duplicadas:",
      dados.duplicated().sum())

print("\nValores ausentes por coluna:")
print(dados.isnull().sum())

print("\nInformações da base original:")
dados.info()

# ==========================================
# 3. REALIZAR OS TRATAMENTOS
# ==========================================

dados_tratados = dados.copy()

# Substituir o símbolo "-" por valor ausente
dados_tratados = dados_tratados.replace(
    r"^\s*-\s*$",
    pd.NA,
    regex=True
)

# Converter as colunas de data
dados_tratados["data_inicio"] = pd.to_datetime(
    dados_tratados["data_inicio"],
    format="%d/%m/%Y",
    errors="coerce"
)

dados_tratados["data_fim"] = pd.to_datetime(
    dados_tratados["data_fim"],
    format="%d/%m/%Y",
    errors="coerce"
)

# Converter as colunas para inteiro
colunas_inteiras = [
    "carga_horaria",
    "matriculas_aptas"
]

# Converter as datas para o formato brasileiro
for coluna in ["data_inicio", "data_fim"]:
    dados[coluna] = pd.to_datetime(
        dados[coluna],
        errors="coerce",
        dayfirst=True
    ).dt.strftime("%d/%m/%Y")

for coluna in colunas_inteiras:
    dados_tratados[coluna] = pd.to_numeric(
        dados_tratados[coluna],
        errors="coerce"
    ).astype("Int64")

# ==========================================
# 4. CONFERÊNCIA DEPOIS DO TRATAMENTO
# ==========================================

print("\n" + "=" * 45)
print("CONFERÊNCIA DEPOIS DO TRATAMENTO")
print("=" * 45)

print("Quantidade de registros:", dados_tratados.shape[0])
print("Quantidade de colunas:", dados_tratados.shape[1])
print("Total de valores ausentes:",
      dados_tratados.isnull().sum().sum())
print("Total de linhas duplicadas:",
      dados_tratados.duplicated().sum())

print("\nValores ausentes por coluna:")
print(dados_tratados.isnull().sum())

print("\nInformações da base tratada:")
dados_tratados.info()

# ==========================================
# 5. COMPARAÇÃO ANTES X DEPOIS
# ==========================================

comparacao = pd.DataFrame({
    "Antes": [
        dados.shape[0],
        dados.shape[1],
        dados.isnull().sum().sum(),
        dados.duplicated().sum()
    ],
    "Depois": [
        dados_tratados.shape[0],
        dados_tratados.shape[1],
        dados_tratados.isnull().sum().sum(),
        dados_tratados.duplicated().sum()
    ]
}, index=[
    "Quantidade de registros",
    "Quantidade de colunas",
    "Valores ausentes",
    "Duplicidades"
])
display(dados[["data_inicio", "data_fim"]].head())
print("\nCOMPARAÇÃO FINAL:")
display(comparacao)
```
